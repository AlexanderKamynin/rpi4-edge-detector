```python
from pathlib import Path

from edge_detector.scripts.evaluation import (
    evaluate_predictions,
    predict_coco,
    print_results,
)


PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

WEIGHTS_DIR = PROJECT_ROOT / "weights" / "state_dict"
PREDICTIONS_DIR = PROJECT_ROOT / "predictions"

IMGSZ = 640
CONF = 0.001
IOU = 0.7
MAX_DET = 300
DEVICE = "cpu"

PREDICTIONS_DIR.mkdir(parents=True, exist_ok=True)
```

```python
GT_JSON = PROJECT_ROOT / "dataset" / "val_coco.json"
IMAGES_DIR = PROJECT_ROOT / "dataset" / "images"

DETECTORS = {
    "InvertedResidual": WEIGHTS_DIR / "custom_shufflenet.pt",
    "MLPBlock + PartialConv": WEIGHTS_DIR / "custom_fasternet.pt",
    "RepVGG": WEIGHTS_DIR / "custom_repvgg.pt",
    "C2f": WEIGHTS_DIR / "custom_c2f.pt",
    "ShuffleNetV2 x0.5": WEIGHTS_DIR / "baseline_shufflenetv2_x0_5.pt",
    "FasterNet T0": WEIGHTS_DIR / "baseline_fasternet_t0.pt",
    "YOLOv8n": WEIGHTS_DIR / "baseline_yolov8n.pt",
    "YOLOv6n": WEIGHTS_DIR / "baseline_yolov6n.pt",
}
```

```python
all_results = {}

for index, (detector_name, checkpoint_path) in enumerate(DETECTORS.items(), start=1):
    prediction_path = PREDICTIONS_DIR / f"{checkpoint_path.stem}.json"

    print(f"[{index}/{len(DETECTORS)}] {detector_name}")

    predict_coco(
        checkpoint_path=checkpoint_path,
        gt_json=GT_JSON,
        images_dir=IMAGES_DIR,
        save_path=prediction_path,
        imgsz=IMGSZ,
        conf=CONF,
        iou=IOU,
        max_det=MAX_DET,
        device=DEVICE,
    )

    results = evaluate_predictions(
        gt_json=GT_JSON,
        pred_json=prediction_path,
    )

    all_results[detector_name] = results

    print_results(results)
    print()
```

```python
detector_name = "FasterNet T0"
checkpoint_path = DETECTORS[detector_name]
prediction_path = PREDICTIONS_DIR / "baseline_fasternet_t0.json"

predict_coco(
    checkpoint_path=checkpoint_path,
    gt_json=GT_JSON,
    images_dir=IMAGES_DIR,
    save_path=prediction_path,
    imgsz=IMGSZ,
    conf=CONF,
    iou=IOU,
    max_det=MAX_DET,
    device=DEVICE,
)

results = evaluate_predictions(
    gt_json=GT_JSON,
    pred_json=prediction_path,
)

print_results(results)
```
